## Условие

$$\Large \frac{∂u}{∂t} = \alpha_x\frac{∂^2u}{∂x^2} + \alpha_y\frac{∂^2u}{∂y^2} + f(t,x,y)$$

$$\Large f(t,x,y) = 1 + m\cos(\omega t) \ \ для\ точек\ (x, y) \in Ω - контур\ области$$

$$\Large f(t,x,y) = 0 \ \ для\ прочих\ точек$$

## Моделирование

```python
import math as mt
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import copy
import random
import plotly.express as px
import matplotlib.animation as animation
from IPython.display import Image
```

### Класс двухмерной области

```python
class Jupyter_grid():
  def __init__(self):
    self.x_size = 21
    self.y_size = 14
    # self.array = np.zeros([self.y_size, self.x_size], dtype='d')
    self.null_cells = np.zeros([self.y_size, self.x_size], dtype='i')
    self.fill_null_cells()

  def fill_null_cells(self):
    self.null_cells[0][0:8] = 1
    self.null_cells[0][15:21] = 1
    self.null_cells[1][0:4] = 1
    self.null_cells[1][18:21] = 1
    self.null_cells[2][0:3] = 1
    self.null_cells[2][20:21] = 1
    self.null_cells[3][0:2] = 1
    self.null_cells[4][0:1] = 1
    self.null_cells[5][0:1] = 1
    self.null_cells[9][20:21] = 1
    self.null_cells[10][19:21] = 1
    self.null_cells[11][0:1] = 1
    self.null_cells[11][18:21] = 1
    self.null_cells[12][0:2] = 1
    self.null_cells[12][17:21] = 1
    self.null_cells[13][0:5] = 1
    self.null_cells[13][14:21] = 1

  def is_outside_cell(self, y, x):
    if self.null_cells[y][x] == 1:
      return True
    return False

  def is_border(self, y, x):
    if (y - 1 < 0 or self.is_outside_cell(y-1, x)):
      return True
    if (x - 1 < 0 or self.is_outside_cell(y, x-1)):
      return True
    if (y + 1 >= self.y_size or self.is_outside_cell(y+1, x)):
      return True
    if (x + 1 >= self.x_size or self.is_outside_cell(y, x+1)):
      return True
    return False
```

### Класс модели

```python
class Model:
  def __init__(self, seasonality, frequency, outside_type, iterations):
    self.iterations = iterations
    self.tau = 0.001
    self.h_x = 0.01
    self.h_y = 0.01
    self.a_x = 0.1
    self.a_y = 0.1
    self.b = [-1/24, 2/3, -5/4, 2/3, -1/24]
    self.d = [-1/3, 1.5, -3, 11/6]
    self.m = seasonality
    self.w = frequency
    self.outside_type = outside_type
    self.outside_fun = lambda t: 1 + self.m* mt.sin(self.w*t)
    self.time = np.linspace(0, self.tau * self.iterations, self.iterations)

    self.jupyter_grid = Jupyter_grid()
    self.start_num = 4
    self.u = np.zeros([self.iterations, self.jupyter_grid.y_size, self.jupyter_grid.x_size], dtype='d')
    self.smooth_distribution()
    self.set_null_cells_null()

  def fill_random(self):
    max_temp = 100
    for i in range(self.start_num):
      self.u[i] = np.random.random((self.jupyter_grid.y_size, self.jupyter_grid.x_size))*max_temp

  def smooth_distribution(self):
    max_temp = 100
    min_temp = 0
    step = 1
    for iter in range(self.start_num):
      self.u[iter,0,:] = np.linspace(max_temp,min_temp,self.jupyter_grid.x_size)
      self.u[iter,:,0] = np.linspace(max_temp,min_temp,self.jupyter_grid.y_size)
      max_temp = max_temp - step
      min_temp = min_temp + step
      for i in range(1, self.jupyter_grid.y_size):
        for j in range(1, self.jupyter_grid.x_size):
          self.u[iter][i][j] = (self.u[iter][i-1][j] + self.u[iter][i][j-1])/2

  def start(self):
    for iter in range(self.start_num, self.iterations):
      for i in range(self.jupyter_grid.y_size):
        for j in range(self.jupyter_grid.x_size):
          if self.jupyter_grid.is_outside_cell(i, j) == True:
            self.u[iter][i][j] = 0
            continue
          fun = 0
          if self.jupyter_grid.is_border(i, j) == True:
            fun = self.outside_fun(self.time[iter])

          x_2derivative = self.get_x_2derivative(iter - 1, i, j)
          y_2derivative = self.get_y_2derivative(iter - 1, i, j)

          self.u[iter][i][j] = (self.tau*(self.a_x * x_2derivative + self.a_y * y_2derivative + fun) - self.d[0]*self.u[iter-3][i][j] - self.d[1]*self.u[iter-2][i][j] - self.d[2]*self.u[iter-1][i][j])/self.d[3]

  def set_null_cells_null(self):
    for iter in range(self.start_num):
      for i in range(self.jupyter_grid.y_size):
        for j in range(self.jupyter_grid.x_size):
          if self.jupyter_grid.is_outside_cell(i, j) == True:
            self.u[iter][i][j] = 0

  def get_x_2derivative(self, iter, y, x):
    if self.outside_type == 'cold':
      return self.get_x_2derivative_cold(iter, y, x)
    if self.outside_type == 'equal':
      return self.get_x_2derivative_equal(iter, y, x)

  def get_y_2derivative(self, iter, y, x):
    if self.outside_type == 'cold':
      return self.get_y_2derivative_cold(iter, y, x)
    if self.outside_type == 'equal':
      return self.get_y_2derivative_equal(iter, y, x)

  def get_x_2derivative_equal(self, iter, y, x):
    result = 0
    if x - 2 >= 0 and self.jupyter_grid.is_outside_cell(y, x - 2)==False:
      result += self.b[0]*self.u[iter][y][ x - 2]
    else:
      result += self.b[0]*self.u[iter][y][x]
    if x - 1 >= 0 and self.jupyter_grid.is_outside_cell(y, x - 1)==False:
      result += self.b[1]*self.u[iter][y][x - 1]
    else:
      result += self.b[1]*self.u[iter][y][x]

    if x >= 0 and self.jupyter_grid.is_outside_cell(y, x)==False:
      result += self.b[2]*self.u[iter][y][x]

    if x + 1 < self.jupyter_grid.x_size and self.jupyter_grid.is_outside_cell(y, x+1)==False:
      result += self.b[3]*self.u[iter][y][x + 1]
    else:
      result += self.b[3]*self.u[iter][y][x]
    if x + 2 < self.jupyter_grid.x_size and self.jupyter_grid.is_outside_cell(y, x+2)==False:
      result += self.b[4]*self.u[iter][y][x + 2]
    else:
      result += self.b[4]*self.u[iter][y][x]

    result = result/self.h_x
    return result

  def get_y_2derivative_equal(self, iter, y, x):
    result = 0
    if y - 2 >= 0 and self.jupyter_grid.is_outside_cell(y - 2, x)==False:
      result += self.b[0]*self.u[iter][y- 2][x]
    else:
      result += self.b[0]*self.u[iter][y][x]
    if y - 1 >= 0 and self.jupyter_grid.is_outside_cell(y - 1, x)==False:
      result += self.b[1]*self.u[iter][y - 1][x]
    else:
      result += self.b[1]*self.u[iter][y][x]
    if y >= 0 and self.jupyter_grid.is_outside_cell(y, x)==False:
      result += self.b[2]*self.u[iter][y][x]
    if y + 1 < self.jupyter_grid.y_size and self.jupyter_grid.is_outside_cell(y+1, x)==False:
      result += self.b[3]*self.u[iter][y+1][x]
    else:
      result += self.b[3]*self.u[iter][y][x]
    if y + 2 < self.jupyter_grid.y_size and self.jupyter_grid.is_outside_cell(y+2, x)==False:
      result += self.b[4]*self.u[iter][y+2][x]
    else:
      result += self.b[4]*self.u[iter][y][x]
    result = result/self.h_y
    return result

  def get_x_2derivative_cold(self, iter, y, x):
    result = 0
    if x - 2 >= 0 and self.jupyter_grid.is_outside_cell(y, x - 2)==False:
      result += self.b[0]*self.u[iter][y][ x - 2]
    if x - 1 >= 0 and self.jupyter_grid.is_outside_cell(y, x - 1)==False:
      result += self.b[1]*self.u[iter][y][x - 1]
    if x >= 0 and self.jupyter_grid.is_outside_cell(y, x)==False:
      result += self.b[2]*self.u[iter][y][x]
    if x + 1 < self.jupyter_grid.x_size and self.jupyter_grid.is_outside_cell(y, x+1)==False:
      result += self.b[3]*self.u[iter][y][x + 1]
    if x + 2 < self.jupyter_grid.x_size and self.jupyter_grid.is_outside_cell(y, x+2)==False:
      result += self.b[4]*self.u[iter][y][x + 2]
    result = result/self.h_x
    return result

  def get_y_2derivative_cold(self, iter, y, x):
    result = 0
    if y - 2 >= 0 and self.jupyter_grid.is_outside_cell(y - 2, x)==False:
      result += self.b[0]*self.u[iter][y- 2][x]
    if y - 1 >= 0 and self.jupyter_grid.is_outside_cell(y - 1, x)==False:
      result += self.b[1]*self.u[iter][y - 1][x]
    if y >= 0 and self.jupyter_grid.is_outside_cell(y, x)==False:
      result += self.b[2]*self.u[iter][y][x]
    if y + 1 < self.jupyter_grid.y_size and self.jupyter_grid.is_outside_cell(y+1, x)==False:
      result += self.b[3]*self.u[iter][y+1][x]
    if y + 2 < self.jupyter_grid.y_size and self.jupyter_grid.is_outside_cell(y+2, x)==False:
      result += self.b[4]*self.u[iter][y+2][x]
    result = result/self.h_y
    return result
```

### Моделирование

```python
model = Model(1,1,"equal", 10000)
model.start()
```

### Сходимость
#### Устойчивость по начальным данным

$$\|u^{n+1}\| \le M_1\|\phi\|, ∀ t^n \in \omega^t - узлы\ сетки\ по\ t$$

```python
min_idx = 2
max_idx = 9999
norms = [np.linalg.norm(model.u[element]) for element in range(min_idx, max_idx)]
fig = plt.figure(figsize=(9, 7))
ax = fig.subplots(1,1)
ax.plot(norms)
ax.set_xlabel("шаг")
ax.set_ylabel("Норма функции от шага")
ax.grid(True, which='both')
ax.minorticks_on()
```

По графику можно сделать вывод, что система устойчива по начальным данным.

#### Аппроксимация
При разложении в ряд Тейлора производных для составлени численной схемы, мы отбрасываем ошибку $O(h^4)$ для рассчётов по $x$ и $y$, и ошибку $O(\tau ^3)$ для рассчётов по времени.

Тогда можем сказать, что будет апроксимация 3-го порядка по времени и 4-го по координате.

#### Сходимость

По теориме П. Лакса - В.С. Рябенького решение будет сходится, если разностная задача устойчива и апроксимирует дифференциальную задачу.

Отсюда решение сходится с 3-м порядком по времени и 4-м по координате.
